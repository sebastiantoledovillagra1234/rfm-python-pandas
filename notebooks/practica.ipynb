```python
import pandas as pd

df = pd.read_excel("../data/Modelo_RFM.xlsx", sheet_name="RFM")
df.head()
```

```python
print("Distribución de clientes por segmento:")
print(df["SEGMENTO"].value_counts())
print()

print("Resumen de monto por segmento:")
resumen = df.groupby("SEGMENTO")["MONTO M$"].agg(["mean", "sum", "count"]).round(0)
print(resumen)
```

```python
segmento_objetivo = "Estrella"
df["SEGMENTO"] = df["SEGMENTO"].str.strip()

clientes_estrella = df[df["SEGMENTO"] == segmento_objetivo]

print(f"Clientes en el segmento '{segmento_objetivo}': {len(clientes_estrella)}")
clientes_estrella[["MONTO M$", "FRECUENCIA", "RECENCIA"]].head()
```
